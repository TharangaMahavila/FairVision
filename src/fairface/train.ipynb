```python
import os
import time
import sys
import torch
import torch.nn as nn
import torch.optim as optim

from data_loader import get_data_loaders
from model import FairVisionCNN

def run_training(epochs=5, batch_size=64, learning_rate=0.001):
    # Optimize acceleration choice for macOS systems (MPS) or NVIDIA (CUDA)
    device = torch.device("mps" if torch.backends.mps.is_available() else ("cuda" if torch.cuda.is_available() else "cpu"))
    print(f"Execution processing configured on device backend: {device}")

    # Fetch modular data pipelines
    train_loader, val_loader = get_data_loaders(batch_size=batch_size)

    # Instantiate model
    model = FairVisionCNN(num_classes=9).to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)

    print("Starting optimization loops...")
    for epoch in range(epochs):
        start_time = time.time()
        model.train()
        
        running_loss = 0.0
        correct_predictions = 0
        total_samples = 0
        
        for batch_idx, batch in enumerate(train_loader):
            images = batch["images"].to(device)
            labels = batch["ages"].to(device)
            
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item() * images.size(0)
            _, preds = torch.max(outputs, 1)
            correct_predictions += torch.sum(preds == labels.data)
            total_samples += images.size(0)
            
            if (batch_idx + 1) % 100 == 0:
                print(f"Epoch [{epoch+1}/{epochs}] | Batch [{batch_idx+1}/{len(train_loader)}] | Target Loss: {loss.item():.4f}")
                
        epoch_loss = running_loss / total_samples
        epoch_acc = correct_predictions.float() / total_samples
        duration = time.time() - start_time
        
        print(f"Epoch {epoch+1} Completed | Metric Loss: {epoch_loss:.4f} | Accuracy Score: {epoch_acc:.2%} | Time: {duration:.1f}s")
        
    # Persist baseline checkpoint files inside the nested package folder structure
    output_path = os.path.join("src", "fairface", "model_weights.pth")
    torch.save(model.state_dict(), output_path)
    print(f"System training successfully resolved. Model weights saved to: {output_path}")

if __name__ == "__main__":
    run_training(epochs=5)
```
